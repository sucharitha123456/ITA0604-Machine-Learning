```python
import pandas as pd

# Step 1: Create the training dataset
data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast': ['Same', 'Same', 'Change', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

# Step 2: Convert data into a DataFrame
df = pd.DataFrame(data)

print("Training Dataset:")
print(df.to_string(index=False))

# Step 3: Initialize the hypothesis
hypothesis = ['0'] * (len(df.columns) - 1)

# Step 4: Apply FIND-S algorithm
for index, row in df.iterrows():
    if row['EnjoySport'] == 'Yes':
        attributes = row.iloc[:-1].tolist()

        if hypothesis == ['0'] * len(hypothesis):
            hypothesis = attributes.copy()
        else:
            for i in range(len(hypothesis)):
                if hypothesis[i] != attributes[i]:
                    hypothesis[i] = '?'

    print("After example", index + 1, ":", hypothesis)

# Step 5: Display the final hypothesis
print("\nFinal Most Specific Hypothesis:")
print(hypothesis)
```

```python
import pandas as pd

data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast': ['Same', 'Same', 'Change', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
df.to_csv('training_data.csv', index=False)

print("CSV file created successfully!")
print(df)

import pandas as pd
from google.colab import files

# Read the training data from CSV
df = pd.read_csv('training_data.csv')

# Separate attributes and target column
X = df.iloc[:, :-1].values.tolist()
y = df.iloc[:, -1].values.tolist()

# Initialize S and G
n = len(X[0])
S = ['0'] * n
G = [['?'] * n]

def covers(h, x):
    return all(h[i] == '?' or h[i] == x[i] for i in range(len(h)))

def more_general_or_equal(h1, h2):
    return all(a == '?' or a == b for a, b in zip(h1, h2))

def minimal_generalizations(h, x):
    result = h.copy()
    for i in range(len(h)):
        if result[i] == '0':
            result[i] = x[i]
        elif result[i] != x[i]:
            result[i] = '?'
    return result

def minimal_specializations(h, x, domains, S):
    result = []
    for i in range(len(h)):
        if h[i] == '?':
            for value in domains[i]:
                if value != x[i]:
                    candidate = h.copy()
                    candidate[i] = value
                    if any(more_general_or_equal(candidate, s) for s in S if s != ['0'] * len(S)):
                        result.append(candidate)
    return result

# Get possible values for each attribute
domains = [sorted(df.iloc[:, i].unique().tolist()) for i in range(n)]

# Process examples
for index, (x, label) in enumerate(zip(X, y), start=1):
    if label.lower() == 'yes':
        # Remove G hypotheses that do not cover positive example
        G = [g for g in G if covers(g, x)]

        # Generalize S to cover the positive example
        if S == ['0'] * n:
            S = x.copy()
        else:
            S = minimal_generalizations(S, x)

        # Keep only hypotheses at least as specific as a G hypothesis
        G = [g for g in G if more_general_or_equal(g, S)]

    else:
        # Remove S hypotheses that cover negative example
        if covers(S, x):
            S = ['0'] * n

        # Specialize G to exclude the negative example
        new_G = []
        for g in G:
            if covers(g, x):
                new_G.extend(minimal_specializations(g, x, domains, [S]))
            else:
                new_G.append(g)

        # Remove duplicate hypotheses
        G = [list(h) for h in set(tuple(h) for h in new_G)]

        # Remove hypotheses more specific than another G hypothesis
        G = [g for g in G if not any(g != other and more_general_or_equal(g, other) for other in G)]

    print(f"\nExample {index}: {label}")
    print("S boundary:", S)
    print("G boundary:", G)

print("\nFinal Specific Boundary (S):")
print(S)

print("\nFinal General Boundary (G):")
for g in G:
    print(g)

print("\nTraining data:")
print(df)
```
